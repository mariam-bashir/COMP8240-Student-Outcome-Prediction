<b> Remove unusable Rahman GPA records </b>

```python
rahman_valid = rahman_clean[
    (rahman_clean["What was your previous SGPA?"] > 0)
    &
    (rahman_clean["What is your current CGPA?"] > 0)
].copy()

print("Original Rahman rows :", len(rahman_clean))
print("Rows removed         :", len(rahman_clean) - len(rahman_valid))
print("Valid Rahman rows    :", len(rahman_valid))
```

<b> Start the harmonised Rahman dataset </b>

```python
rahman_h = pd.DataFrame()

print("Empty harmonised dataset created.")
```

<b> Add gender </b>

```python
rahman_h["gender"] = (
    rahman_valid["Gender"]
    .str.strip()
)

print(rahman_h["gender"].value_counts())
```

<b> Add study stage </b>

```python
def map_study_stage(value):
    value = int(value)

    if value <= 4:
        return "Early"
    elif value <= 8:
        return "Middle"
    else:
        return "Late"
```

```python
rahman_h["study_stage"] = (
    rahman_valid["Current Semester"]
    .apply(map_study_stage)
)

print(rahman_h["study_stage"].value_counts())
```

<b> Add study preparation </b>

```python
def map_study_preparation(hours):
    if hours <= 1:
        return "0-1 Hour"
    elif hours <= 3:
        return "2-3 Hours"
    else:
        return "More than 3 Hours"
```

```python
rahman_h["study_preparation"] = (
    rahman_valid["How many hour do you study daily?"]
    .apply(map_study_preparation)
)

print(rahman_h["study_preparation"].value_counts())
```

<b> Clean Rahman attendance </b>

```python
rahman_attendance = (
    rahman_valid["Average attendance on class"]
    .replace("94-98", 96)
)

rahman_attendance = pd.to_numeric(
    rahman_attendance,
    errors="coerce"
)

print("Conversion failures:", rahman_attendance.isna().sum())
print(rahman_attendance.describe())
```

<B> Convert attendance into common categories </b>

```python
def map_attendance(value):
    if value < 40:
        return "Below 40%"
    elif value < 60:
        return "40%-59%"
    elif value < 80:
        return "60%-79%"
    else:
        return "80%-100%"
```

```python
rahman_h["attendance"] = (
    rahman_attendance
    .apply(map_attendance)
)

print(rahman_h["attendance"].value_counts())
```

<b> Add extracurricular participation </b>

```python
rahman_h["extracurricular_participation"] = (
    rahman_valid[
        "Are you engaged with any co-curriculum activities?"
    ]
    .str.strip()
)

print(
    rahman_h["extracurricular_participation"]
    .value_counts()
)
```

<b> Add family income </b>

```python
def map_family_income(value):
    if value < 15000:
        return "Low"
    elif value < 30000:
        return "Lower middle"
    elif value <= 50000:
        return "Upper middle"
    else:
        return "High"
```

```python
rahman_h["family_income"] = (
    rahman_valid["What is your monthly family income?"]
    .apply(map_family_income)
)

print(rahman_h["family_income"].value_counts())
```

<b> Add previous academic performance </b>

```python
rahman_h["previous_academic_performance"] = (
    rahman_valid["What was your previous SGPA?"]
)

print(
    rahman_h["previous_academic_performance"]
    .describe()
)
```

<b> Create Rahman academic-performance outcome </b>

Current CGPA is used only to construct the target variable and is not
included as a predictor.

The binary outcome follows the project proposal:

- **Higher Academic Performance:** GPA >= 3.0
- **Lower Academic Performance:** GPA < 3.0

This is an operational Higher/Lower academic-performance definition,
not an institutional Pass/Fail threshold. 

```python
rahman_h["outcome"] = (
    rahman_valid["What is your current CGPA?"]
    .apply(
        lambda x: "Higher" if x >= 3.0 else "Lower"
    )
)

print(rahman_h["outcome"].value_counts())
```

<b> Add source dataset </b>

```python
rahman_h["source_dataset"] = "Rahman"

print(rahman_h["source_dataset"].value_counts())
```

<b> Validate the finished Rahman dataset </b>

```python
print("Rahman harmonised shape:", rahman_h.shape)

print("\nColumns:")
for col in rahman_h.columns:
    print("-", col)

print("\nMissing values:")
print(rahman_h.isnull().sum())

print("\nOutcome distribution:")
print(rahman_h["outcome"].value_counts())

rahman_h.head()
```

<b>Start Hasan harmonised dataset</b>

```python
hasan_h = pd.DataFrame()

print("Empty Hasan harmonised dataset created.")
```

<b>Add gender</b>

```python
hasan_h["gender"] = (
    hasan_cse["Gender"]
    .str.strip()
)

print(hasan_h["gender"].value_counts())
```

<b> Convert Hasan semester to numeric </b>

```python
hasan_semester = (
    hasan_cse["Semester"]
    .str.extract(r"(\d+)")[0]
    .astype(int)
)

print(hasan_semester.value_counts().sort_index())
```

<b>Add Hasan study stage</b>

```python
hasan_h["study_stage"] = (
    hasan_semester
    .apply(map_study_stage)
)

print(hasan_h["study_stage"].value_counts())
```

<b>Add attendance</b>

```python
hasan_h["attendance"] = (
    hasan_cse["Attendance"]
    .str.strip()
)

print(hasan_h["attendance"].value_counts())
```

<b>Add extracurricular participation</b>

```python
hasan_h["extracurricular_participation"] = (
    hasan_cse["Extra"]
    .str.strip()
)

print(
    hasan_h["extracurricular_participation"]
    .value_counts()
)
```

<b>Add family income</b>

```python
income_map = {
    "Low (Below 15,000)": "Low",
    "Lower middle (15,000-30,000)": "Lower middle",
    "Upper middle (30,000-50,000)": "Upper middle",
    "High (Above 50,000)": "High"
}

hasan_h["family_income"] = (
    hasan_cse["Income"]
    .map(income_map)
)

print(hasan_h["family_income"].value_counts())

print(
    "Unmapped income values:",
    hasan_h["family_income"].isna().sum()
)
```

<b>Add previous academic performance</b>

```python
hasan_h["previous_academic_performance"] = (
    hasan_cse["Last"]
)

print(
    hasan_h["previous_academic_performance"]
    .describe()
)
```

<b>Create Hasan academic-performance outcome </b>

Overall GPA is used only to construct the target variable and is not
included as a predictor.

The same common outcome definition is applied:

- **Higher Academic Performance:** GPA >= 3.0
- **Lower Academic Performance:** GPA < 3.0

```python
hasan_h["outcome"] = (
    hasan_cse["Overall"]
    .apply(
        lambda x: "Higher" if x >= 3.0 else "Lower"
    )
)

print(hasan_h["outcome"].value_counts())
```

<b> study_preparation </b>

```python
hasan_h["study_preparation"] = (
    hasan_cse["Preparation"]
    .str.strip()
)

print(hasan_h["study_preparation"].value_counts())
```

<b>Add source</b>

```python
hasan_h["source_dataset"] = "Hasan"

print(hasan_h["source_dataset"].value_counts())
```

<b>Validate Hasan</b>

```python
print("Hasan harmonised shape:", hasan_h.shape)

print("\nColumns:")
for col in hasan_h.columns:
    print("-", col)

print("\nMissing values:")
print(hasan_h.isnull().sum())

print("\nOutcome distribution:")
print(hasan_h["outcome"].value_counts())

hasan_h.head()
```

<b> Put both datasets in exactly the same column order </b>

```python
common_columns = [
    "gender",
    "study_stage",
    "study_preparation",
    "attendance",
    "extracurricular_participation",
    "family_income",
    "previous_academic_performance",
    "outcome",
    "source_dataset"
]

rahman_h = rahman_h[common_columns]
hasan_h = hasan_h[common_columns]

print("Rahman columns identical:", rahman_h.columns.tolist() == common_columns)
print("Hasan columns identical :", hasan_h.columns.tolist() == common_columns)
```

<b>Final check before combining</b>

```python
print("Rahman shape:", rahman_h.shape)
print("Hasan shape :", hasan_h.shape)

print("\nRahman missing values:", rahman_h.isnull().sum().sum())
print("Hasan missing values :", hasan_h.isnull().sum().sum())
```

<b>Combine the datasets</b>

```python
constructed_df = pd.concat(
    [rahman_h, hasan_h],
    ignore_index=True
)

print("Combined dataset shape:", constructed_df.shape)
```

<b>Validate source contribution</b>

```python
print(constructed_df["source_dataset"].value_counts())
```

<b>Check final outcome distribution</b>

```python
print(constructed_df["outcome"].value_counts())

print("\nOutcome percentages:")
print(
    constructed_df["outcome"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

<b>Check all harmonised categories</b>

```python
categorical_columns = [
    "gender",
    "study_stage",
    "study_preparation",
    "attendance",
    "extracurricular_participation",
    "family_income"
]

for col in categorical_columns:
    print("\n" + "=" * 60)
    print(col)
    print("=" * 60)
    print(constructed_df[col].value_counts())
```

<b>Final missing-value check</b>

```python
print("Total missing values:")
print(constructed_df.isnull().sum())

print("\nTotal missing values in dataset:")
print(constructed_df.isnull().sum().sum())
```
